```python
#extraction of digits
n = 5873
num = n
while num > 0:
  last_digit = num % 10
  print(last_digit)
  num = num//10
```

```python
#count digits
n = 5873
count = 0
while n > 0:
  n = n // 10
  count = count + 1
  print(count)
```

```python
#check palindrome = palindrome is something which reads the same from backward and forward
n = 1234
num = n
result = 0
while num > 0:
  ld = num % 10
  result = (result*10) + ld
  num = num // 10
n == result
```

```python
#armstrong numbers = the sum of each digit raised to the power of the number of digits equals to original numbers.
n = 153
num = n
total = 0
nod = len (str(n))
while num > 0:
  ld = num % 10
  total = total + (ld ** nod)
  num = num // 10
total == n
```
